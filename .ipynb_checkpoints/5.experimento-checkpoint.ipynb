```python
import pandas as pd
import numpy as np
import time
import math
import os
import xgboost as xgb

import lightgbm as lgb
from lightgbm import LGBMRanker

from sklearn.linear_model import LogisticRegression
from sklearn.metrics import roc_auc_score
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import Pipeline

from tqdm import tqdm
print("fim")
tqdm.pandas()
```

```python
def load_data (split):
    df_train = pd.read_csv(f"./splits/{split}/train_pares.csv")
    df_val = pd.read_csv(f"./splits/{split}/val_pares.csv")
    df_test = pd.read_csv(f"./splits/{split}/test_pares.csv")

    return df_train, df_val, df_test
```

```python
def ranking_metrics(group):
    group_id = group.name

    group = group.sort_values("score", ascending=False)

    labels = group["label"].values
    relevant_positions = np.where(labels == 1)[0]

    n_relevant = len(relevant_positions)

    #se não há relevantes, estoura erro
    if n_relevant == 0:
        #print(f"Nenhum parente encontrado id={group_id}")
        return pd.Series({
            "id": group_id,
            "recall@1": np.nan,
            "recall@5": np.nan,
            "recall@10": np.nan,
            "recall@15": np.nan,
            "recall@20": np.nan,
            "mrr": np.nan,
            "count": np.nan
        })

    # Recall@1
    recall_at_1 = int(labels[0] == 1) / n_relevant

    # Recall@5
    recall_at_5 = labels[:5].sum() / n_relevant

    # Recall@10
    recall_at_10 = labels[:10].sum() / n_relevant

    # Recall@10
    recall_at_15 = labels[:15].sum() / n_relevant

    # Recall@10
    recall_at_20 = labels[:20].sum() / n_relevant

    # MRR
    first_relevant_rank = relevant_positions[0] + 1
    mrr = 1.0 / first_relevant_rank

    return pd.Series({
        "id": group_id,
        "recall@1": recall_at_1,
        "recall@5": recall_at_5,
        "recall@10": recall_at_10,
        "recall@15": recall_at_15,
        "recall@20": recall_at_20,
        "mrr": mrr,
        "count": len(labels)
    })
```

```python
#nova célula 4 com função inferencia 
def inferencia(label, df_test, split_path, features, model, mode="ranker"):
    """Realiza inferência e calcula métricas de ranking"""
    
    X_test = df_test[features].copy()
    
    if mode == "heuristic":
        cols_to_fill = ["age_score", "addr_score", "occ_score"]
        X_test[cols_to_fill] = X_test[cols_to_fill].fillna(0)
        scores = (
            0.50 * X_test["age_score"] +
            0.40 * X_test["addr_score"] +
            0.10 * X_test["occ_score"]
        )
    
    elif mode == "classifier":
        cols_to_fill = ["age_score", "addr_score", "occ_score", "addr_exato"]
        X_test[cols_to_fill] = X_test[cols_to_fill].fillna(0)
        scores = model.predict_proba(X_test)[:, 1]
    
    elif mode == "ranker":
        scores = model.predict(X_test)
    
    df_test_copy = df_test.copy()
    df_test_copy["score"] = scores
    
    df_results = df_test_copy.groupby(["ID"]).apply(ranking_metrics)
    df_results = df_results.reset_index(drop=True)
    df_results.to_csv(f"{split_path}/{label}.csv", index=False)
    
    return df_results



```

```python
#one hot encoding 

def create_one_hot_encoding(df):
    relation_categories = [
        'Spouse', 'Biological child', 'Adopted child', 'Child-in-law',
        'Parent-in-law', 'Parent', 'Grandchild', 'Grandparent', 'Sibling',
        'Nephew/niece', 'Stepchild', 'Stepparent', 'Sibling-in-law'
    ]
    
    #coloca 0 ou 1 nas categorias 
    for category in relation_categories:
        col_name = f"is_{category.replace(' ', '').replace('/', '_')}"
        df[col_name] = (df['RELATED'] == category).astype(int)
    
    return df

def prepare_features(df_train, df_valid, df_test):
    #prepara as features para os três datasets
    df_train = create_one_hot_encoding(df_train)
    df_valid = create_one_hot_encoding(df_valid)
    df_test = create_one_hot_encoding(df_test)
    
    return df_train, df_valid, df_test
```

```python
#célula nova do xgboost
def train_xgb_ranker(df_train, df_valid, features):
    """Treina modelo XGBoost Ranker"""
    
    df_train = df_train.sort_values("ID")
    df_valid = df_valid.sort_values("ID")
    
    X_train = df_train[features]
    Y_train = df_train["label"]

    X_valid = df_valid[features]
    Y_valid = df_valid["label"]
   
    group_train = (
        df_train
        .groupby("ID")
        .size()
        .values
    )
    
    group_valid = (
        df_valid
        .groupby("ID")
        .size()
        .values
    )

    model = xgb.XGBRanker(
        objective="rank:ndcg",
        eval_metric="ndcg@10",
        n_estimators=300,
        max_depth=6,
        learning_rate=0.05,
        subsample=0.8,
        colsample_bytree=0.8,
        tree_method="hist",
        random_state=42,
        n_jobs=-1
    )

    model.fit(
        X_train,
        Y_train,
        group=group_train,
        eval_set=[(X_valid, Y_valid)],
        eval_group=[group_valid],
        verbose=20
    )

    return model
```

```python
#célula nova do lgbm

def train_lgbm_ranker(df_train, df_valid, features):
    """Treina modelo LightGBM Ranker"""
    
    df_train = df_train.sort_values("ID")
    df_valid = df_valid.sort_values("ID")
    
    X_train = df_train[features]
    Y_train = df_train["label"]

    X_valid = df_valid[features]
    Y_valid = df_valid["label"]
    
    #grupos para ranking (precisa ser lista para LightGBM)
    group_train = df_train.groupby("ID").size().tolist()
    group_valid = df_valid.groupby("ID").size().tolist()

    model = LGBMRanker(
        objective="lambdarank",
        metric="ndcg",
        n_estimators=300,
        learning_rate=0.05,
        max_depth=-1,
        num_leaves=31,
        subsample=0.8,
        colsample_bytree=0.8,
        random_state=42,
        n_jobs=-1
    )
    
    model.fit(
        X_train,
        Y_train,
        group=group_train,
        eval_set=[(X_valid, Y_valid)],
        eval_group=[group_valid],
        eval_at=[1, 5, 10],
        callbacks=[
            lgb.log_evaluation(period=20)
        ]
    )

    return model
```

```python
#célula nova do reg_log 

def train_reg_log_classifier(df_train, df_valid, features):
    """Treina modelo de Regressão Logística"""
    
    #colunas sem one code 
    cols_to_fill = [
        "age_score",
        "addr_score",
        "occ_score",
        "addr_exato"
    ]
    
    # dados de treino
    X_train = df_train[features].copy()
    X_train[cols_to_fill] = X_train[cols_to_fill].fillna(0)
    Y_train = df_train["label"]
    
    # dados de validação
    X_valid = df_valid[features].copy()
    X_valid[cols_to_fill] = X_valid[cols_to_fill].fillna(0)
    Y_valid = df_valid["label"]
    
    #calcular peso das classes (balanceamento)
    n_pos = (Y_train == 1).sum()
    n_neg = (Y_train == 0).sum()
    
    class_weight = {
        0: 1.0,
        1: n_neg / n_pos  # Dar mais peso aos positivos (são minoria)
    }
    

    model = LogisticRegression(
        penalty="l2",
        solver="lbfgs",
        max_iter=3000,
        class_weight=class_weight,
        n_jobs=-1
    )
    
    #pipeline com normalização
    scaler = StandardScaler()
    pipeline = Pipeline([
        ("scaler", scaler),
        ("clf", model)
    ])
    
    #treino
    pipeline.fit(X_train, Y_train)
    
    #validar
    y_valid_proba = pipeline.predict_proba(X_valid)[:, 1]
    auc = roc_auc_score(Y_valid, y_valid_proba)
    print(f"  Validation AUC: {auc:.4f}")
    
    return pipeline
```

```python
#célula nova com heurística, xgboost e lgbm

OUTPUT_PATH = "output"
os.makedirs(OUTPUT_PATH, exist_ok=True)


for split in range(1, 6):
    print(f"\n{'='*50}")
    print(f"SPLIT {split}")
    print(f"{'='*50}\n")
    
    split_path = OUTPUT_PATH+f"/{split}"
    os.makedirs(split_path, exist_ok=True)
    
    # Carregar dados
    print(f"Loading Data {split}")
    df_train, df_val, df_test = load_data(split)
    
    # Shuffle dos dados de treino
    print(f"Shuffling train data {split}")
    df_train = df_train.sample(frac=1, random_state=42).reset_index(drop=True)
    
    # Criar one-hot encoding
    print(f"Creating one-hot encoding {split}")
    df_train = pd.get_dummies(df_train, columns=['RELATED'], prefix='is')
    df_val = pd.get_dummies(df_val, columns=['RELATED'], prefix='is')
    df_test = pd.get_dummies(df_test, columns=['RELATED'], prefix='is')
    
    # Garantir que todos os dataframes tenham as mesmas colunas
    all_columns = list(set(df_train.columns) | set(df_val.columns) | set(df_test.columns))
    for col in all_columns:
        if col not in df_train.columns:
            df_train[col] = 0
        if col not in df_val.columns:
            df_val[col] = 0
        if col not in df_test.columns:
            df_test[col] = 0
    
    # Atualizar features com as novas colunas is_*
    is_columns = [col for col in df_train.columns if col.startswith('is_')]
    features_rank = ['age_score', 'addr_score', 'addr_exato', 'occ_score'] + is_columns


    # ==================== XGBoost Ranker ====================
    label = "xgb_ranker"
    print(f"Training XGBoost Ranker {split}")
    xgb_model = train_xgb_ranker(df_train, df_val, features_rank)
    
    print(f"Inference XGBoost Ranker {split}")
    inferencia(label, df_test, split_path, features_rank, xgb_model, mode="ranker")

    # ==================== LightGBM Ranker ====================
    label = "lgbm_ranker"
    print("Training model LGBM", split)
    lgbm_model = train_lgbm_ranker(df_train, df_val, features_rank)
    print(f"Inference LightGBM Ranker {split}")
    inferencia(label, df_test, split_path, features_rank, lgbm_model, mode="ranker")

    # ==================== Regressão Logística  ====================
     
    features_reg_log = features_rank + [
        'missing_age',
        'missing_addr',
        'missing_occ',
    ]
    label = "reg_log"
    print(f"Training model Regressão Logistica {split}")
    reg_log_model = train_reg_log_classifier(df_train, df_val, features_reg_log)
    print(f"Inference model Regressão Logistica {split}")
    inferencia(label, df_test, split_path, features_reg_log, reg_log_model, mode="classifier")

    # ==================== Heuristica  ====================
    features_heuristic = [
        "age_score",
        "addr_score",
        "occ_score",
    ]
    label = "heuristica"
    print("Inference heuristica", split)
    inferencia(label, df_test, split_path, features_heuristic, None, mode="heuristic")

    # ==================== LightGBM Ranker (COMENTADO) ====================
    # label = "lgbm_ranker"
    # print(f"Training LightGBM Ranker {split}")
    # lgbm_model = train_lgbm_ranker(df_train, df_val, features_rank)
    # print(f"Inference LightGBM Ranker {split}")
    # inferencia(label, df_test, split_path, features_rank, lgbm_model, mode="ranker")
    
    # ==================== Regressão Logística (COMENTADO) ====================
    # label = "reg_log"
    # print(f"Training model Regressão Logistica {split}")
    # reg_log_model = train_reg_log_classifier(df_train, df_val, features_rank)
    # print(f"Inference model Regressão Logistica {split}")
    # inferencia(label, df_test, split_path, features_rank, reg_log_model, mode="classifier")
    
    # XGBoost (comentado por enquanto)
    #label = "xgboost_ranker"
    #print("Training model XGBoost", split)
    # TREINAR XGBoost com features_rank
    #print("Inference model XGBoost", split)
    # INFERENCIA XGBoost

print("fim")
```

```python
LABELS = ["heuristica", "reg_log", "xgb_ranker", "lgbm_ranker"]

# Descomentar linha abaixo quando os outros metodos estiverem prontos.
#LABELS = ["heuristica", "reg_log", "xgboost_ranker", "lgbm_ranker"]

SPLITS = [1,2,3,4,5]
OUTPUT_PATH = "output" 
metrics_to_print = ["recall@1", "recall@5", "recall@10", "recall@15", "recall@20", "mrr"]
result = []
for label in LABELS:

    for split in SPLITS:
        split_path = f"{OUTPUT_PATH}/{split}"
        file_path = f"{split_path}/{label}.csv"
        df_results = pd.read_csv(file_path)
        
        metrics = df_results[metrics_to_print].mean()
        metrics["split"] = split
        metrics["label"] = label
        result.append(metrics)

df = pd.DataFrame(result)



print(f"\n{'='*60}")
print("RESULTADOS CONSOLIDADOS")
print(f"{'='*60}\n")
print(df.to_string())

# Mostrar média por método
print(f"\n{'='*60}")
print("MÉDIA POR MÉTODO")
print(f"{'='*60}\n")
df_avg = df.groupby("label")[metrics_to_print].mean()
print(df_avg.to_string())

```

```python
#pedi pro chat gerar um código de verificação pro embaralhamento
def verificar_embaralhamento_treino(split, n_samples=10):
    """Verifica se o embaralhamento foi aplicado nos dados de treino"""
    
    print(f"\nVERIFICAÇÃO DE EMBARALHAMENTO - SPLIT {split}\n")
    
    # Carregar dados originais (sem embaralhar)
    df_train_original = pd.read_csv(f"./splits/{split}/train_pares.csv")
    
    # Simular o processo do loop: carregar e embaralhar
    df_train_shuffled = pd.read_csv(f"./splits/{split}/train_pares.csv")
    df_train_shuffled = df_train_shuffled.sample(frac=1, random_state=RANDOM_SEED).reset_index(drop=True)
    
    print("ORDEM ORIGINAL (primeiras linhas):")
    print(df_train_original[['ID', 'label', 'RELATED']].head(n_samples))
    
    print("\nORDEM APÓS EMBARALHAMENTO (primeiras linhas):")
    print(df_train_shuffled[['ID', 'label', 'RELATED']].head(n_samples))
    
    print("\nESTATÍSTICAS:")
    print(f"  Total de registros original: {len(df_train_original):,}")
    print(f"  Total de registros embaralhado: {len(df_train_shuffled):,}")
    print(f"  IDs únicos original: {df_train_original['ID'].nunique():,}")
    print(f"  IDs únicos embaralhado: {df_train_shuffled['ID'].nunique():,}")
    print(f"  Labels positivos original: {df_train_original['label'].sum():,}")
    print(f"  Labels positivos embaralhado: {df_train_shuffled['label'].sum():,}")
    
    # Verificar se a ordem mudou
    ids_originais = df_train_original['ID'].head(20).tolist()
    ids_embaralhados = df_train_shuffled['ID'].head(20).tolist()
    ordem_mudou = ids_originais != ids_embaralhados
    
    print(f"\n  Primeiros 20 IDs originais: {ids_originais[:5]}...")
    print(f"  Primeiros 20 IDs embaralhados: {ids_embaralhados[:5]}...")
    print(f"\n  Ordem mudou? {ordem_mudou}")
    
    if ordem_mudou:
        print("  EMBARALHAMENTO CONFIRMADO")
    else:
        print("  ATENÇÃO: A ordem parece não ter mudado")
    
    # Verificar que val e test NÃO foram embaralhados
    print("\nVERIFICANDO QUE VAL E TEST NÃO FORAM EMBARALHADOS:")
    
    df_valid = pd.read_csv(f"./splits/{split}/val_pares.csv")
    df_test = pd.read_csv(f"./splits/{split}/test_pares.csv")
    
    print(f"  Validation - Primeiros IDs: {df_valid['ID'].head(5).tolist()}")
    print(f"  Test - Primeiros IDs: {df_test['ID'].head(5).tolist()}")
    print(f"  Val e Test permanecem na ordem original\n")


# Executar verificação no split 1
print("VERIFICANDO EMBARALHAMENTO DO TREINO")
verificar_embaralhamento_treino(split=1, n_samples=15)

# Verificar distribuição de labels após embaralhamento
print("\nDISTRIBUIÇÃO DE LABELS APÓS EMBARALHAMENTO")

for split in SPLITS:
    df_train = pd.read_csv(f"./splits/{split}/train_pares.csv")
    df_train_shuffled = df_train.sample(frac=1, random_state=RANDOM_SEED).reset_index(drop=True)
    
    print(f"\nSplit {split}:")
    print(f"  Labels positivos: {df_train_shuffled['label'].sum():,} ({df_train_shuffled['label'].mean()*100:.2f}%)")
    print(f"  Labels negativos: {(df_train_shuffled['label']==0).sum():,} ({(1-df_train_shuffled['label'].mean())*100:.2f}%)")

print("\nVERIFICAÇÕES CONCLUÍDAS")
```
