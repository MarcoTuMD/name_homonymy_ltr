```python
def save_local_test_data(split):

    # executar query
    query = f"""
    SELECT 
        MIN(id) AS min_id,
        MAX(id) AS max_id
    FROM trilhas.homonimo_teste_{split}
    """
    cursor = conn.cursor()
    cursor.execute(query)

    # pegar resultado
    min_id, max_id = cursor.fetchall()[0]
    cursor.close()
    HIVE_PAGE_SIZE = 100

    last_id = min_id - 1
    all_dfs = []

    total_bar = math.ceil((max_id - min_id) / HIVE_PAGE_SIZE)
    pbar = tqdm(total=total_bar)

    all_results = []
    cont = 1
    while last_id < max_id:
        print(f"{last_id} -- {last_id + HIVE_PAGE_SIZE}")
        df_test = pd.read_sql(f"""
            select
                id,
                is_mae,
                is_pai,
                is_filho,
                is_avos,
                is_neto,
                is_irmao,
                is_tio,
                is_sobrinho,
                is_primo,
                is_conjuge,
                is_sogro,
                label,
                age_score,
                missing_idade,
                address_score_ml_base,
                prof_score_ml_base,
                missing_profissao,
                missing_endereco
            from trilhas.homonimo_teste_{split}
            WHERE (id > {last_id}) and (id <= {last_id + HIVE_PAGE_SIZE})
            ORDER BY id
        """, conn)

        if df_test.empty:
            break

        df_test.to_csv(f"test_data_{split}/{last_id}.csv", index=False)
        last_id = df_test["id"].max()
        pbar.update(1)

    pbar.close()
```

```python
def train_reg_log_classifier (df_train, df_valid, features):
    cols_to_fill = [
        "age_score",
        "address_score_ml_base",
        "prof_score_ml_base",
        "addr_lev_logradouro",
        "addr_ext_cidade",
        "addr_ext_estado",
    ]

    X_train = df_train[features]
    X_train[cols_to_fill] = X_train[cols_to_fill].fillna(0)
    Y_train = df_train["label"]

    X_valid = df_valid[features]
    X_valid[cols_to_fill] = X_valid[cols_to_fill].fillna(0)
    Y_valid =  df_valid["label"]

    scaler = StandardScaler()

    n_pos = (Y_train == 1).sum()
    n_neg = (Y_train == 0).sum()
    
    class_weight = {
        0: 1.0,
        1: n_neg / n_pos
    }
    
    model = LogisticRegression(
        penalty="l2",
        solver="lbfgs",
        max_iter=3000,
        class_weight=class_weight,
        n_jobs=-1
    )
    
    pipeline = Pipeline([
        ("scaler", scaler),
        ("clf", model)
    ])

    pipeline.fit(X_train, Y_train)

    y_valid_proba = pipeline.predict_proba(X_valid)[:, 1]
    auc = roc_auc_score(Y_valid, y_valid_proba)

    print("Validation AUC:", auc)

    return pipeline
```

```python
def train_xgb_ranker (df_train, df_valid, features):

    df_train = df_train.sort_values("id")
    df_valid = df_valid.sort_values("id")
    
    X_train = df_train[features]
    Y_train = df_train["label"]

    X_valid = df_valid[features]
    Y_valid =  df_valid["label"]
   
    group_train = (
        df_train
        .groupby("id")
        .size()
        .values
    )
    
    group_valid = (
        df_valid
        .groupby("id")
        .size()
        .values
    )

    model = xgb.XGBRanker(
        objective="rank:ndcg",
        eval_metric="ndcg@10",
        n_estimators=300,
        max_depth=6,
        learning_rate=0.05,
        subsample=0.8,
        colsample_bytree=0.8,
        tree_method="hist",
        random_state=42,
        n_jobs=-1
    )

    model.fit(
        X_train,
        Y_train,
        group=group_train,
        eval_set=[(X_valid, Y_valid)],
        eval_group=[group_valid],
        verbose=20
    )

    return model
```

```python
def train_lgbm_ranker (df_train, df_valid, features):
    df_train = df_train.sort_values("id")
    df_valid = df_valid.sort_values("id")
    
    X_train = df_train[features]
    Y_train = df_train["label"]

    X_valid = df_valid[features]
    Y_valid =  df_valid["label"]
    
    group_train = df_train.groupby("id").size().tolist()
    group_valid = df_valid.groupby("id").size().tolist()

    model = LGBMRanker(
        objective="lambdarank",
        metric="ndcg",
        n_estimators=300,
        learning_rate=0.05,
        max_depth=-1,
        num_leaves=31,
        subsample=0.8,
        colsample_bytree=0.8,
        random_state=42,
        n_jobs=-1
    )
    
    model.fit(
        X_train,
        Y_train,
        group=group_train,
        eval_set=[(X_valid, Y_valid)],
        eval_group=[group_valid],
        eval_at=[1, 5, 10],
        callbacks=[
            lgb.log_evaluation(period=20)
        ]
    )

    return model
```

```python
def ranking_metrics(group):
    group_id = group.name

    group = group.sort_values("score", ascending=False)

    labels = group["label"].values
    relevant_positions = np.where(labels == 1)[0]

    n_relevant = len(relevant_positions)

    # Se não há relevantes, estoura erro
    if n_relevant == 0:
        #print(f"Nenhum parente encontrado id={group_id}")
        return pd.Series({
            "id": group_id,
            "recall@1": np.nan,
            "recall@5": np.nan,
            "recall@10": np.nan,
            "recall@15": np.nan,
            "recall@20": np.nan,
            "mrr": np.nan,
            "count": np.nan
        })

    # Recall@1
    recall_at_1 = int(labels[0] == 1) / n_relevant

    # Recall@5
    recall_at_5 = labels[:5].sum() / n_relevant

    # Recall@10
    recall_at_10 = labels[:10].sum() / n_relevant

    # Recall@10
    recall_at_15 = labels[:15].sum() / n_relevant

    # Recall@10
    recall_at_20 = labels[:20].sum() / n_relevant

    # MRR
    first_relevant_rank = relevant_positions[0] + 1
    mrr = 1.0 / first_relevant_rank

    return pd.Series({
        "id": group_id,
        "recall@1": recall_at_1,
        "recall@5": recall_at_5,
        "recall@10": recall_at_10,
        "recall@15": recall_at_15,
        "recall@20": recall_at_20,
        "mrr": mrr,
        "count": len(labels)
    })
```

```python
def inferencia_batch (df_test, features, model, mode):

    X_test = df_test[features]

    #classifier
    if mode == "classifier":
        cols_to_fill = [
            "age_score",
            "address_score_ml_base",
            "prof_score_ml_base",
            "addr_lev_logradouro",
            "addr_ext_cidade",
            "addr_ext_estado",
        ]
        X_test[cols_to_fill] = X_test[cols_to_fill].fillna(0)

        scores = model.predict_proba(X_test)[:, 1]

    if mode == "ranker":
        scores = model.predict(X_test)

    df_test["score"] = scores

    metrics_per_id = df_test.groupby(["id"]).apply(ranking_metrics)
    return metrics_per_id.reset_index(drop=True)
```

```python
def inferencia (label, split, features, model, mode="ranker"):
    all_results = []

    split_path = f"test_data_{split}"
    files = [
        f for f in os.listdir(split_path)
        if os.path.isfile(os.path.join(split_path, f))
    ]
    pbar = tqdm(total=len(files))
    for file in files:
        df_test = pd.read_csv(f"{split_path}/{file}")
        
        if df_test.empty:
            break
        batch_metrics = inferencia_batch(df_test, features, model, mode)
        all_results.extend(batch_metrics.to_dict("records"))

        pbar.update(1)

    pbar.close()

    df_results = pd.DataFrame(all_results)
    df_results.to_csv(f"result--{label}--{split}.csv", index=False)
    df_results = df_results[df_results["mrr"].notna()]
    
    # Lista de métricas
    metrics_to_print = ["recall@1", "recall@5", "recall@10", "recall@15", "recall@20", "mrr"]

    metrics = df_results[metrics_to_print].mean()

    print(label, split, "\n", metrics, "\n")
```
