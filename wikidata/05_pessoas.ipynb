```python
import os
import json

input_file = 'wikidata_pessoa_homonimo.jsonl'
output_distinct = 'wikidata_distinct_pessoa_dados.jsonl'

print(f"Lendo '{input_file}' para extração de pessoas únicas...")

pessoas_unicas = {}
total_linhas_originais = 0

with open(input_file, 'r', encoding='utf-8') as f:
    for line in f:
        total_linhas_originais += 1
        
        if not line.strip():
            continue
            
        row = json.loads(line)
        
        p1 = row.get('Dados_pessoa_1')
        if p1 and isinstance(p1, dict) and 'id' in p1:
            pessoas_unicas[p1['id']] = p1
            
        p2 = row.get('Dados_pessoa_2')
        if p2 and isinstance(p2, dict) and 'id' in p2:
            pessoas_unicas[p2['id']] = p2
            
        lista_homonimos = row.get('Lista_homonimos', [])
        if isinstance(lista_homonimos, list):
            for p_h in lista_homonimos:
                if p_h and isinstance(p_h, dict) and 'id' in p_h:
                    pessoas_unicas[p_h['id']] = p_h

print(f"Salvando resultados em '{output_distinct}'...")

with open(output_distinct, 'w', encoding='utf-8') as f_out:
    for pessoa in pessoas_unicas.values():
        f_out.write(json.dumps(pessoa, ensure_ascii=False) + '\n')

def format_bytes(size):
    power = 2**10
    n = 0
    power_labels = {0 : 'B', 1: 'KB', 2: 'MB', 3: 'GB', 4: 'TB'}
    while size > power and n < 4:
        size /= power
        n += 1
    return f"{size:.2f} {power_labels[n]}"

try:
    size_orig = os.path.getsize(input_file)
    size_dist = os.path.getsize(output_distinct)
except OSError:
    size_orig = 0
    size_dist = 0

print(f"\n↳ Nome: {input_file}")
print(f"↳ Total de linhas lidas: {total_linhas_originais}")
print(f"↳ Tamanho: {format_bytes(size_orig)}\n")

print(f"↳ Nome: {output_distinct}")
print(f"↳ Total de pessoas únicas: {len(pessoas_unicas)}")
print(f"↳ Tamanho: {format_bytes(size_dist)}\n")

print("Processo concluído.")
```
