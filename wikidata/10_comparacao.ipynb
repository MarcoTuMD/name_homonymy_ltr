```python
import pandas as pd
from scipy.stats import wilcoxon

lgbm_results = {
    'MRR':       [0.9290, 0.9276, 0.9283, 0.9285, 0.9276],
    'Recall@1':  [0.8827, 0.8798, 0.8811, 0.8813, 0.8804],
    'Recall@5':  [0.9847, 0.9851, 0.9847, 0.9853, 0.9844],
    'Recall@10': [0.9922, 0.9925, 0.9923, 0.9924, 0.9923],
    'Recall@15': [0.9946, 0.9949, 0.9947, 0.9946, 0.9946],
    'Recall@20': [0.9961, 0.9961, 0.9959, 0.9957, 0.9956]
}

xgb_results = {
    'MRR':       [0.9275, 0.9270, 0.9274, 0.9273, 0.9269],
    'Recall@1':  [0.8805, 0.8789, 0.8794, 0.8794, 0.8793],
    'Recall@5':  [0.9840, 0.9848, 0.9847, 0.9847, 0.9840],
    'Recall@10': [0.9919, 0.9924, 0.9922, 0.9923, 0.9922],
    'Recall@15': [0.9946, 0.9950, 0.9947, 0.9945, 0.9946],
    'Recall@20': [0.9959, 0.9962, 0.9959, 0.9957, 0.9955]
}

metricas = ['MRR', 'Recall@1', 'Recall@5', 'Recall@10', 'Recall@15', 'Recall@20']
estatisticas_w = []
p_valores = []

for metrica in metricas:
    modelo_a = xgb_results[metrica]
    modelo_b = lgbm_results[metrica]
    
    # hipótese de "Modelo B (LGBM) > Modelo A (XGBoost)", por isso o alternative='greater'
    # wilcoxon(x, y, alternative='greater') testa se a distribuição de x - y é maior que 0
    stat, p_value = wilcoxon(x=modelo_b, y=modelo_a, alternative='greater')
    
    estatisticas_w.append(round(stat, 4))
    p_valores.append(round(p_value, 4))

tabela_wilcoxon = pd.DataFrame({
    'Métrica': metricas,
    'W': estatisticas_w,
    'p_valor': p_valores
}).set_index('Métrica').T

print("XGBoost Ranker vs LGBM Ranker")
display(tabela_wilcoxon)
```
