```python
import json
import pandas as pd
import os

input_file = 'wikidata_distinct_pessoa_dados.jsonl'
output_file = 'embeddings.csv'

profissoes_set = set()
enderecos_set = set()
locais_nascimento_set = set()
nacionalidades_set = set()

try:
    with open(input_file, 'r', encoding='utf-8') as f:
        for line in f:
            try:
                data = json.loads(line)

                if 'profissao' in data:
                    profissoes_set.update(f"Occupation: {p}" for p in data['profissao'])

                if 'endereço' in data:
                    enderecos_set.update(f"Residence: {e}" for e in data['endereço'])

                if 'local_nascimento' in data:
                    locais_nascimento_set.update(f"Birthplace: {l}" for l in data['local_nascimento'])

                if 'nacionalidade' in data:
                    nacionalidades_set.update(f"Nationality: {n}" for n in data['nacionalidade'])

            except json.JSONDecodeError:
                print(f"Erro ao ler uma linha específica do arquivo.")
                continue
    
    df_com_prefixo = pd.DataFrame({
        'Profissoes': pd.Series(list(profissoes_set)),
        'Enderecos': pd.Series(list(enderecos_set)),
        'Locais_Nascimento': pd.Series(list(locais_nascimento_set)),
        'Nacionalidades': pd.Series(list(nacionalidades_set))
    })
    
    print(f"Profissões únicas encontradas: {len(profissoes_set)}")
    print(f"Endereços únicos encontrados: {len(enderecos_set)}")
    print(f"Locais de nascimento únicos encontrados: {len(locais_nascimento_set)}")
    print(f"Nacionalidades únicas encontradas: {len(nacionalidades_set)}")

    df_com_prefixo.to_csv(output_file, index=False, encoding='utf-8')
    print(f"Arquivo '{output_file}' criado com as chaves.")

except FileNotFoundError:
    print(f"Erro: O arquivo '{input_file}' não foi encontrado no diretório.")
```
